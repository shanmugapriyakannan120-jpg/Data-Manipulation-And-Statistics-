```python
OPERATION ON ARRAY

1. Element-wise Arithmetic Operations:** Performs arithmetic operations like addition, subtraction, multiplication, and division on each element of an array.

2. Scalar Operation:** Performs an operation between every array element and a single value.

3. Comparison and Logical Operation:** Compares array elements and returns Boolean values (True or False).

4. **Aggregation Operation:** Calculates summary values like sum, mean, minimum, maximum, and standard deviation.

5. **Shape and Restructuring Operation:** Changes the shape or arrangement of an array without changing its data.

6. **Mathematical Function:** Applies mathematical functions like square root, power, and exponential to array elements.

7. **Broadcasting:** Allows arrays of different shapes to perform operations together.

8. **Matrix Operation:** Performs matrix calculations like multiplication, transpose, and inverse.

9. **Sorting and Searching:** Arranges array elements in order and finds specific values or their positions.
```
