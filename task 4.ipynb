# task 2
HANDLING MISSING VALUE 


```python
print("Missing Values:")
print(df.isnull().sum())
```

```python
missing_cols = df.isnull().sum()
print(
    missing_cols[missing_cols > 0]
)
```

To fill missing values, we used:

Mean: Used for numerical data; replaces missing values with the average.
Median: Used for numerical data with outliers; replaces missing values with the middle value.
Mode: Used for categorical/text data; replaces missing values with the most frequent value.

```python
# fill missing value in city using mode

df['City']=df['City'].fillna(
    df['City'].mode()[0])
```

```python
#fillna(): Used to replace missing values.
#mode()[0]: Returns the most frequent value in the column.
#inplace=True: Applies the changes directly to the original DataFrame.
```

```python
#Fill Missing Values in Product Name Using Mode

df['Quantity_Available']=df['Quantity_Available'].fillna(
    df['Quantity_Available'].mean()
    )
```

```python
#Fill Missing Values in Units Sold Using Mean

df['Units_Sold']=df['Units_Sold'].fillna(
    df['Units_Sold'].mean()
    )
```

```python
df['Product_Name']=df['Product_Name'].fillna(
    df['Product_Name'].mode()[0]
)
```

```python
#Fill Missing Values in Revenue Using Mean

df['Revenue']=df['Revenue'].fillna(
    df['Revenue'].mean()
    )
```

```python
#Fill Missing Values in Supplier Using Mode

df['Supplier']=df['Supplier'].fillna(
    df['Supplier'].mode()[0]
    )
```

```python
#Fill Missing Values in Customer satisfaction Using Mode

df['Customer_Satisfaction']=df['Customer_Satisfaction'].fillna(
    df['Customer_Satisfaction'].mean()
    )
```

```python
#Fill Missing Values in Delivery Time Using Mean

df['Delivery_Time']=df['Delivery_Time'].fillna(
    df['Delivery_Time'].mean()
    )
```

```python
df.dtypes
```

```python
df['Shipping_Cost'] = pd.to_numeric(
    df['Shipping_Cost'],
    errors='coerce'
)

print(df['Shipping_Cost'].dtypes)
```

```python
#Fill Missing Values in Shipping Cost Using Mean

df['Shipping_Cost']=df['Shipping_Cost'].fillna(
    df['Shipping_Cost'].mean()
    )
```
